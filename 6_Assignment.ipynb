```python
# Q11. Write a Python program to calculate the area of a rectangle using length and breadth given by the user. 

l = float(input('Write length of rectangle: '))
b = float(input('Write breadth of rectangle: '))

area = lambda l,b : l*b
print('Area of Rectangle: ',area(l,b))
```

```python
# Q12. Write a Python program to find the quotient and remainder when one number is divided by another number. 

num_1 = float(input('Write first number: '))
num_2 = float(input('Write second number: '))

quotient = lambda num_1,num_2 : num_1 // num_2
reminder = lambda num_1,num_2 : num_1 % num_2

print('Quotient: ',quotient(num_1,num_2))
print('Reminder: ',reminder(num_1,num_2))
```
