```python
# Q3. Create a list of 5 colors. Insert "Black" at the second position using insert(). 

color = ['Red', 'Green', 'Yellow', 'Purple', 'Blue']

color.insert(1,'Black')
print(color)
```

```python
# Q15. Take age as a string and height as a string. Convert age into an integer and height into a float.

age = '55'
height = '6.5'
print('Befor: ')
print('Age = ',age,': ',type(age))
print('Height = ',height,': ',type(height))

age = int(age)
height = float(height)
print("")
print('After: ')
print('Age = ',age,': ',type(age))
print('Height = ',height,': ',type(height))
```
